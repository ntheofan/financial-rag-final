```python
from pathlib import Path
import os
import json
import random
import warnings

import numpy as np
import pandas as pd
from dotenv import load_dotenv
```

```python
CURRENT_DIR = Path.cwd()

if CURRENT_DIR.name == "notebooks":
    BASE_DIR = CURRENT_DIR.parent
else:
    BASE_DIR = CURRENT_DIR

NOTEBOOKS_DIR = BASE_DIR / "notebooks"
DATA_DIR = BASE_DIR / "data"
RAW_DIR = DATA_DIR / "raw"
INTERIM_DIR = DATA_DIR / "interim"
PROCESSED_DIR = DATA_DIR / "processed"

PDFS_DIR = RAW_DIR / "pdfs"
MARKDOWN_DIR = INTERIM_DIR / "markdown"
PARSED_JSON_DIR = INTERIM_DIR / "parsed_json"
CLEANED_DIR = INTERIM_DIR / "cleaned"

CHUNKS_DIR = PROCESSED_DIR / "chunks"
EMBEDDINGS_DIR = PROCESSED_DIR / "embeddings"
RETRIEVAL_DIR = PROCESSED_DIR / "retrieval_results"
EVALUATION_DIR = PROCESSED_DIR / "evaluation"

CONFIG_DIR = BASE_DIR / "configs"
OUTPUTS_DIR = BASE_DIR / "outputs"
FIGURES_DIR = OUTPUTS_DIR / "figures"
TABLES_DIR = OUTPUTS_DIR / "tables"
LOGS_DIR = OUTPUTS_DIR / "logs"

print("BASE_DIR:", BASE_DIR)
print("DATA_DIR:", DATA_DIR)
print("CONFIG_DIR:", CONFIG_DIR)
print("OUTPUTS_DIR:", OUTPUTS_DIR)
```

```python
# Write your code here
required_dirs = [
    NOTEBOOKS_DIR,
    DATA_DIR,
    RAW_DIR,
    PDFS_DIR,
    INTERIM_DIR,
    MARKDOWN_DIR,
    PARSED_JSON_DIR,
    CLEANED_DIR,
    PROCESSED_DIR,
    CHUNKS_DIR,
    EMBEDDINGS_DIR,
    RETRIEVAL_DIR,
    EVALUATION_DIR,
    CONFIG_DIR,
    OUTPUTS_DIR,
    FIGURES_DIR,
    TABLES_DIR,
    LOGS_DIR,
]

missing_dirs = [str(path) for path in required_dirs if not path.exists()]

if missing_dirs:
    print("Missing directories found:")
    for d in missing_dirs:
        print("-", d)
else:
    print("All required directories exist.")
```

```python
for path in required_dirs:
    path.mkdir(parents=True, exist_ok=True)

print("All directories are now ensured.")
```

```python
config_path = CONFIG_DIR / "config.json"

if config_path.exists():
    with open(config_path, "r", encoding="utf-8") as f:
        config = json.load(f)
    print("config.json loaded successfully.")
else:
    config = {}
    print("config.json not found. Using empty config.")

config
```

```python
def load_json_config(path: Path):
    if path.exists():
        with open(path, "r", encoding="utf-8") as f:
            return json.load(f)
    return {}

chunking_config = load_json_config(CONFIG_DIR / "chunking_config.json")
retrieval_config = load_json_config(CONFIG_DIR / "retrieval_config.json")
evaluation_config = load_json_config(CONFIG_DIR / "evaluation_config.json")

print("chunking_config:", chunking_config)
print("retrieval_config:", retrieval_config)
print("evaluation_config:", evaluation_config)
```

```python
def save_json(data, path: Path, indent: int = 2):
    path.parent.mkdir(parents=True, exist_ok=True)
    with open(path, "w", encoding="utf-8") as f:
        json.dump(data, f, ensure_ascii=False, indent=indent)
```

```python
def load_json(path: Path):
    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)
```

```python
def save_dataframe(df: pd.DataFrame, path: Path, index: bool = False):
    path.parent.mkdir(parents=True, exist_ok=True)

    if path.suffix.lower() == ".csv":
        df.to_csv(path, index=index)
    elif path.suffix.lower() == ".parquet":
        df.to_parquet(path, index=index)
    else:
        raise ValueError("Unsupported file format. Use .csv or .parquet")
```
